# Baseline Profiling Demo

This notebook demonstrates the basic profiling capabilities of PESquared for Python functions.

We'll:
1. Define an intentionally inefficient function
2. Profile it to identify bottlenecks
3. Manually optimize it
4. Compare performance before and after

```python
# Add parent directory to path to import modules
import sys
sys.path.insert(0, '..')

from analyzers.profiler import profile_function, format_profile_summary
from verifiers.verifier import measure_runtime, compare_performance, print_performance_comparison
```

## 1. Define an Inefficient Function

Let's create a function that computes Fibonacci numbers using a naive recursive approach.
This is known to be inefficient due to repeated calculations.

```python
def naive_fibonacci(n):
    """Inefficient recursive Fibonacci implementation."""
    if n <= 1:
        return n
    return naive_fibonacci(n - 1) + naive_fibonacci(n - 2)

# Test it
result = naive_fibonacci(10)
print(f"fibonacci(10) = {result}")
```

## 2. Profile the Function

Now let's profile this function to see where time is being spent.

```python
# Profile the naive implementation
summary = profile_function(naive_fibonacci, 20, n_runs=1)

# Display human-readable summary
print(format_profile_summary(summary, llm_friendly=False))
```

## 3. Manually Create an Optimized Version

Based on our understanding of the problem, we can optimize using dynamic programming (iterative approach).

```python
def optimized_fibonacci(n):
    """Efficient iterative Fibonacci implementation."""
    if n <= 1:
        return n
    
    a, b = 0, 1
    for _ in range(2, n + 1):
        a, b = b, a + b
    return b

# Verify correctness
print(f"naive_fibonacci(20) = {naive_fibonacci(20)}")
print(f"optimized_fibonacci(20) = {optimized_fibonacci(20)}")
```

## 4. Compare Performance

Let's measure and compare the performance of both implementations.

```python
# Compare performance
comparison = compare_performance(
    naive_fibonacci,
    optimized_fibonacci,
    25,  # Using n=25 for reasonable runtime
    n_runs=10
)

print_performance_comparison(comparison)
```

## 5. Another Example: Inefficient List Operations

Let's look at another common performance issue: inefficient list operations.

```python
def inefficient_sum_squares(n):
    """Inefficient: creates full list in memory."""
    return sum([i**2 for i in range(n)])

def efficient_sum_squares(n):
    """Efficient: uses generator expression."""
    return sum(i**2 for i in range(n))

# Compare performance
comparison2 = compare_performance(
    inefficient_sum_squares,
    efficient_sum_squares,
    100000,
    n_runs=100
)

print_performance_comparison(comparison2)
```

## Key Takeaways

1. **Profiling** helps identify where time is being spent
2. **Algorithmic improvements** (like changing from O(2^n) to O(n)) provide the biggest gains
3. **Memory efficiency** (generators vs lists) can also improve performance
4. **Always verify correctness** when optimizing
